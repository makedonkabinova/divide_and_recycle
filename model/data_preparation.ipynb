### Dataset: https://www.kaggle.com/datasets/viswaprakash1990/garbage-detection?select=GARBAGE+CLASSIFICATION

```python
import os

# Root folder where train/test/valid live
ROOT = "garbage_dataset"
DEST = "data_prepared"

# Your classes
classes = ['biodegradable', 'cardboard', 'glass', 'metal', 'paper', 'plastic']

# Create destination folder structure
for split in ['train', 'valid', 'test']:
    for cls in classes:
        os.makedirs(os.path.join(DEST, split, cls), exist_ok=True)

images_by_class = {cls: [] for cls in classes}

# Loop through train/test/valid folders
for split in ['train', 'test', 'valid']:
    split_path = os.path.join(ROOT, split, "images")

    if not os.path.isdir(split_path):
        continue

    # Loop through all images in this split
    for filename in os.listdir(split_path):
        file_path = os.path.join(split_path, filename)

        # Skip folders we already created
        if os.path.isdir(file_path):
            continue

        # Detect class from filename prefix
        filename_lower = filename.lower()

        matched_class = None
        for cls in classes:
            if filename_lower.startswith(cls):
                matched_class = cls
                break

        # If no match, skip or print warning
        if matched_class is None:
            print(f"Could not detect class for: {filename}")
            continue

        images_by_class[matched_class].append(file_path)
```

```python
import random
import shutil

# Now split and copy
for cls, file_list in images_by_class.items():
    random.shuffle(file_list)

    total = len(file_list)
    train_end = int(total * 0.8)
    valid_end = int(total * 0.9)  # 80% train, next 10% valid, last 10% test

    train_files = file_list[:train_end]
    valid_files = file_list[train_end:valid_end]
    test_files  = file_list[valid_end:]

    # Copy Train
    for f in train_files:
        shutil.copy(f, os.path.join(DEST, "train", cls))

    # Copy Valid
    for f in valid_files:
        shutil.copy(f, os.path.join(DEST, "valid", cls))

    # Copy Test
    for f in test_files:
        shutil.copy(f, os.path.join(DEST, "test", cls))

    print(f"{cls}: {len(train_files)} train, {len(valid_files)} valid, {len(test_files)} test")
```
